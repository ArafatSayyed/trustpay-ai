```python
# Create a sample transaction
sample_transaction = pd.DataFrame([{
    "step": 1,
    "amount": 1000,
    "oldbalanceOrg": 5000,
    "newbalanceOrig": 4000,
    "oldbalanceDest": 2000,
    "newbalanceDest": 3000,
    "isFlaggedFraud": 0,
    "type_CASH_IN": 0,
    "type_CASH_OUT": 1,
    "type_DEBIT": 0,
    "type_PAYMENT": 0,
    "type_TRANSFER": 0
}])

print("Sample Transaction:")
print(sample_transaction)
```

```python
# Make prediction
prediction = model.predict(sample_transaction)[0]
probability = model.predict_proba(sample_transaction)[0][1]

if prediction == 1:
    result = "FRAUDULENT TRANSACTION"
else:
    result = "NON-FRAUDULENT TRANSACTION"

print("Prediction:", result)
print("Fraud Probability:", probability)
```
